## Problem 1: Downloading Stock Market Data

We will create a function called `get_data()` which uses the `yf.download()` function from the yfinance package to download stock market data.

This will provide a list containing the five company stock tickers. `period="5d"` gives us five days of historical data, while `interval="1h"` requests hourly data.

[yfinance Docs](https://ranaroussi.github.io/yfinance/reference/api/yfinance.download.html)

[Datetime - strftime()](https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior)

```python
# Download five days of hourly stock market data
def get_data():
    df = yf.download(
        ["NVDA", "AVGO", "ARM", "AMD", "IBM"],
        period="5d",
        interval="1h"
    )

    # Generate a timestamp for the filename
    now = dt.datetime.now()
    timestamp = now.strftime("%Y%m%d-%H%M%S")

    # Save the downloaded data as a CSV file
    filename = f"data/{timestamp}.csv"
    df.to_csv(filename)

    return df

# Call the function and display the downloaded data
stock_data = get_data()
stock_data
```

### Generating a Timestamp + Saving Data as a CSV File

We use `dt.datetime.now()` to retrieve the current date and time when the function is executed.

`strftime()` is used to format the date and time into the required filename format of `YYYYMMDD-HHmmss`.

We then use `df.to_csv()` to save the downloaded stock market data as a CSV file inside the data folder.

Using a timestamp in the filename ensures that each download has a unique filename, preventing previous downloads from being overwritten.

[Pandas - df.to_csv()](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.to_csv.html)
